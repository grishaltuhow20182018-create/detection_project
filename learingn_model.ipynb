```python
from ultralytics import YOLO

# Загружаем быструю модель
model_fast = YOLO('yolov8n.pt')

# Используем встроенный датасет (YOLO скачает его сама без Kaggle)
results_fast = model_fast.train(
    data='coco128.yaml',
    epochs=30,
    imgsz=640
)
```

```python
print("Starting Accurate Model Training...")

model_accurate = YOLO('yolov8m.pt')
model_accurate.train(
    data='coco128.yaml',
    epochs=30,
    imgsz=640,
    name='accurate_model_run' # Папка для сохранения: runs/detect/accurate_model_run
)
```
